16) Write a program to check if a year is leap year or not.

If a year is divisible by 4 then it is leap year but if the year is century year like 2000, 1900, 2100 then it must be divisible by 400.

```python
Year=int(input("Enter a year to check whether it is a leap year or not: "))

if((Year%400==0) or (Year%4==0 and Year%100!=0)):
    print(f'{Year} is a leap year ')
else:
    print(f'{Year} is not a leap year')
    
```

13) Take input of age of 3 people by user and determine oldest and youngest among them.

```python
person1_age=int(input("Enter Age for person 1 "))
person2_age=int(input("Enter Age for person 2 "))
person3_age=int(input("Enter Age for person 3 "))

oldest=(max(person1_age,person2_age,person3_age))

youngest=(min(person1_age,person2_age,person3_age))


print(f"The oldest person is {oldest} years old.")
print(f"The youngest person is {youngest} years old.")
```
