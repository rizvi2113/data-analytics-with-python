6. Take the radius of a circle from a variable and calculate its Area and Circumference (use `pi = 3.14159`)

```python
Circle_rad=8
pi=3.14159
Circle_area=pi*Circle_rad**2
Circum=2*pi*Circle_rad
print(f'Area:{Circle_area}')
print(f'Circumference:{Circum}')
```
